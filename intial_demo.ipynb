```python
import requests
import json
import os
from datetime import datetime

base_url = "https://api.sportsdata.io/v3/nfl/stats/json/PlayerGameStatsByWeekFinal"

headers = {
    "Ocp-Apim-Subscription-Key": API_KEY
}

all_data = {}

for week in range(1, 19):
    url = f"{base_url}/2025/{week}"
    
    response = requests.get(url, headers=headers)
    
    print(f"Week {week}: Status {response.status_code}")
    
    if response.status_code == 200:
        all_data[week] = response.json()
        print(f"   Records: {len(all_data[week])}")
    else:
        print(f"   Error: {response.text}")

print("\nFinished!")
print("Weeks collected:", len(all_data))
```

```python
import boto3
import pandas as pd
from io import StringIO

# -----------------------------
# 1. Create some NFL data
# -----------------------------

data = {
    "player": ["Josh Allen", "Lamar Jackson", "Joe Burrow"],
    "position": ["QB", "QB", "QB"],
    "passing_yards": [4306, 3678, 4918],
    "passing_tds": [35, 31, 43],
    "interceptions": [10, 6, 9],
    "rushing_yards": [531, 915, 155],
    "rushing_tds": [12, 4, 2],
    "team_wins": [13, 12, 10]
}

df = pd.DataFrame(data)

print("NFL data:")
print(df)


# -----------------------------
# 2. Connect to Amazon S3
# -----------------------------

s3 = boto3.client("s3")

bucket_name = "YOUR-NFL-MVP-BUCKET"
file_name = "raw/nfl_player_stats.csv"


# -----------------------------
# 3. Convert data to CSV
# -----------------------------

csv_buffer = StringIO()
df.to_csv(csv_buffer, index=False)


# -----------------------------
# 4. Upload data to S3
# -----------------------------

s3.put_object(
    Bucket=bucket_name,
    Key=file_name,
    Body=csv_buffer.getvalue()
)

print("NFL data successfully uploaded to S3!")
print(f"Location: s3://{bucket_name}/{file_name}")
```

```python
import requests
import pandas as pd
import boto3
from io import StringIO

# SportsDataIO API
API_KEY = "YOUR_API_KEY_HERE"

season = 2026
week = 3

url = f"https://api.sportsdata.io/v3/nfl/stats/json/PlayerGameStatsByWeekFinal/{season}/{week}"

headers = {
    "Ocp-Apim-Subscription-Key": API_KEY
}

response = requests.get(url, headers=headers)

print("Status code:", response.status_code)

data = response.json()

print("Number of player records:", len(data))
```
